* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
matrix = ([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
diag_len = min(len(matrix), len(matrix[0]))
mult = 1
for i in range(diag_len):
    value = matrix[i][i]
    if value != 0:
        mult *= value
        
print (mult)        
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
new_x = np.array([2, 2, 2, 3, 3, 3, 5]) 
values = []
counts = []
current = new_x[0]
count = 1
for i in range(1, len(new_x)):
    if new_x[i] == current:
        count += 1
    else:
        values.append(current)
        counts.append(count)
        current = new_x[i]
        count = 1
values.append(current)
counts.append(count)
values = np.array(values)
counts = np.array(counts)

print((values, counts))
```

```python
x = np.array([2, 2, 2, 3, 3, 3, 5, 8 , 8 , 5 ,1])
ind = np.where(np.diff(x) != 0)[0] + 1
blocks = np.concatenate(([0], ind, [len(x)]))
values = x[blocks[:-1]]
counts = np.diff(blocks)
print((values, counts))
```
